# Sync Notion DB with Discogs
Quickly sync existing Collection DB from Notion with Discogs data

### Initialize environment variables

```typescript
import 'dotenv/config';
```

### Exports Discogs data
Data is saved into `files/discogs_collection.json` file.

```typescript
import { exportCollection } from './export-discogs-collection.ts'

await exportCollection({
  consumerKey: Deno.env.get('DISCOGS_CONSUMER_KEY'),
  consumerSecret: Deno.env.get('DISCOGS_CONSUMER_SECRET'),
});
```

### Exports Notion data
Data is saved into `files/notion_collection.json` file.

```typescript
import { exportCollection } from './export-notion-collection.ts'

await exportCollection({
  auth: Deno.env.get('NOTION_AUTH_TOKEN'),
  databaseId: Deno.env.get('NOTION_DATABASE_ID'),
});
```

### Format Notion data (optional)

Formats data to a more readable way, and save it into `files/notion_collection_formatted.json`.

```typescript
import { formatNotionCollection } from './collection-formatter.ts'

formatNotionCollection();
```

### Update Notion DB

* Reads `files/discogs_collection.json` and compares to `files/notion_vinyl_collection.json`
  * **ATTENTION**: Please only run if **both files are up-to-date** to avoid data deletion
* Updates Notion DB with new items from Discogs

```typescript
import { syncDbCollection } from './notion-vinyl-db-utils.ts';

const forceUpdate = false; // Set to true if to update all existing items

await syncDbCollection({
  auth: Deno.env.get('NOTION_AUTH_TOKEN'),
  databaseId: Deno.env.get('NOTION_DATABASE_ID'),
}, forceUpdate);
```
