# Export Discogs collection
Quickly export vinyl collection from Discogs.

```typescript
import 'dotenv/config';
```

### Export data

```typescript
import { exportCollection } from './export-discogs-lib.ts'

await exportCollection({
  consumerKey: Deno.env.get('DISCOGS_CONSUMER_KEY'),
  consumerSecret: Deno.env.get('DISCOGS_CONSUMER_SECRET'),
});
```
